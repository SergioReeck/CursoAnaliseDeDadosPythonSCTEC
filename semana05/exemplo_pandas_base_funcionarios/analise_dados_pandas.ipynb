### Importar Bases de Dados (Excel)
### Criar os DataFrames Referentes a Cada Base

```python
import pandas as pd
```

```python
# Importar Arquivos e Criar DataFrames

df_funcionarios = pd.read_excel('funcionarios.xlsx')
df_historico_salarios = pd.read_excel('historico_salarios.xlsx')
df_setores = pd.read_excel('setores.xlsx')
```

```python
df_funcionarios.head()
```

### Exibir Nome e Salario da Base de Funcionários

```python
print(df_funcionarios[["Nome", "Salario"]])
```

### Filtrar Funcionários com Salários Acima de R$ 5.000

```python
print(df_funcionarios[df_funcionarios["Salario"] > 5000])    
```

### Merge Entre Funcionários e Setores (Juntar Tabelas)

```python
df_funcionarios_setor = df_funcionarios.merge(df_setores, on='IdSetor')
```

```python
df_funcionarios_setor
```

```python
df_funcionarios_setor[["Nome", "Setor", "Salario"]]
```

### Adicionar Novos Funcionários (Concat)

```python
novos_funcionarios = {
    "IdFuncionario": [6, 7],
    "Nome": ["Gabriel", "Helena"],
    "IdSetor": [3, 4],
    "Salario": [4500, 5300],
    "Idade": [31, 38],
    }

df_novos_funcionarios = pd.DataFrame(novos_funcionarios)
```

```python
df_novos_funcionarios
```

```python
df_funcionarios = pd.concat([df_funcionarios, df_novos_funcionarios], ignore_index=True)
```
