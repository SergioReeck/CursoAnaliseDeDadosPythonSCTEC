```python
import pandas as pd
```

```python
dados = {
    "Produto": ["Notebook", "Mouse", "Teclado", "Monitor"],
    "Quantidade": [2, 5, 3, 1],
    "Valor": [3500.00, 80.00, 150.00, 1200.00]
}
```

```python
df = pd.DataFrame(dados)
```

```python
df
```

```python
print("Valor: ")
print(df["Valor"])
print()
print(type(df["Valor"]))
```

```python
print(df[["Produto", "Valor"]])
```

```python
# Filtrando Valores

print(df[df["Valor"] > 200])
```

```python
# Filtrando Valores e Selecionando Colunas

print(df[df["Valor"] >= 150][["Produto", "Valor"]])
```

```python
# Criando uma Nova Coluna com Base nas Informações de Quantidade e Valor

df["Total"] = df["Quantidade"] * df["Valor"]
```

```python
df
```

```python
# Média de Valor

print(df["Valor"].mean())
```

```python
# Máximo e Mínimo

print(df["Valor"].max())
print(df["Valor"].min())
```

```python
# Contagem

print(df["Produto"].count())
```

```python
# Exibir Últimas Linhas

print(df.tail(2))
```
