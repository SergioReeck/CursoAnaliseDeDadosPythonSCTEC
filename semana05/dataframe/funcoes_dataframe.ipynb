### Group By

```python
dados = {
    "Categoria": ["Computador", "Periférico", "Periférico", "Periférico", "Acessório", "Computador", "Hardware", "Hardware"],
    "Produto": ["Notebook", "Mouse", "Teclado", "Monitor", "Mousepad", "MiniPC", "SSD", "Memória RAM"],
    "Quantidade": [2, 5, 3, 1, 5, 2, 3, 3],
    "Valor": [3500.00, 80.00, 150.00, 1200.00, 10.00, 2500.00, 200.00, 190.00]
}
```

```python
import pandas as pd
```

```python
df_produtos = pd.DataFrame(dados)
```

```python
df_produtos.to_excel("exemplo_produtos.xlsx", index=False)
```

```python
df_produtos
```

```python
# Agrupamento para Soma de Valor

print(df_produtos.groupby("Categoria")["Valor"].sum())
```

```python
# Agrupamento por Categoria (Pegando a Quantidade de Produtos Distintos (Nome))

print(df_produtos.groupby("Categoria")["Produto"].count())
```

```python
# Estatística - Agrupamento

print(df_produtos.groupby("Categoria")["Valor"].agg(["sum", "mean", "max", "min"]))
```
