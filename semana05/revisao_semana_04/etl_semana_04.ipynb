```python
# Importação de Bibliotecas/Módulos

import re
from datetime import datetime
```

```python
# Função para Converter o Valor

def converter_valor(valor): # 1.250,50
    valor = valor.replace(".", "") # Saída -> 1250,00
    valor = valor.replace(",", ".") # Saída -> 1250.50

    return float(valor)
```

```python
# Função para Converter Data

def converter_data(data):
    return datetime.strptime(data, "%d/%m/%Y")
```

```python
# Função para Processar Linhas do Relatório 

def processar_linha(linha):

    # Ignorar Cancelados
    if "CANCELADO" in linha:
        return None

    # Encontrar Padrões
    padrao = (
        r"Pedido #(\d+)\s*\|\s*"
        r"Data=(\d{2}/\d{2}/\d{4})\s*\|\s*"
        r"Cliente:\s*(.*?)\s*\|\s*"
        r"Valor:\s*R\$\s*([\d.,]+)"
    )

    resultado = re.search(padrao, linha)

    if resultado is None:
        return None

    pedido, data, cliente, valor = resultado.groups()

    return {
        "pedido": int(pedido),
        "data": converter_data(data),
        "cliente": cliente,
        "valor": converter_valor(valor)
    }
    
```

```python
# Lendo o Arquivo

vendas = []

with open("vendas_julho.txt", "r", encoding="utf-8") as arquivo:
    for linha in arquivo:
        venda = processar_linha(linha)

        if venda:
            vendas.append(venda)

        
```

```python
# Exibindo as vendas

for venda in vendas:
    print(
        venda["pedido"],
        venda["data"].strftime("%d/%m/%Y"),
        venda["cliente"],
        f"R$ {venda["valor"]:.2f}"
    )
```

```python
# Função para Calcular o Total de Vendas do Mês

def calcular_total(vendas):
    total = 0

    for venda in vendas:
        total += venda["valor"]

    return total
```

```python
print(f"Quantidade de Vendas: {len(vendas)}")
print(f"Valor Total Vendido: R$ {calcular_total(vendas):.2f}")
```
