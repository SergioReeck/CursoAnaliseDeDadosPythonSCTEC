# Exemplo 4 - Filtros e Estatísticas

```python
vendas = pd.Series([150, 100, 180, 200, 80, 350, 400])
```

```python
# Total Vendido

print("Total Vendido: ", vendas.sum())
```

```python
# Total Máximo e Valor Mínimo

print("Máximo: ", vendas.max())
print("Mínimo: ", vendas.min())
```

```python
# Média

print("Média: ", vendas.mean())
```

```python
# Quantidade de Elementos/Linhas

print("Contagem: ", vendas.count())
```

```python
# Comparação

print(vendas > 200)
```

```python
# Filtrar Valor

print(vendas[vendas > 200])
```
