### Pandas

```python
import pandas as pd
```

```python
df_funcionarios = pd.DataFrame({
    "Nome": ["Ana", "Carlos", "Maria"],
    "Salario": [3500.00, 4200.00, 5100.00],
    "Setor": ["RH", "TI", "Financeiro"]
})

print(df_funcionarios)
```

```python
df_funcionarios.head(1)  # Exibe as primeiras linhas do DataFrame
```

```python
df_funcionarios.tail(1)  # Exibe as últimas linhas do DataFrame
```

```python
print(df_funcionarios.describe())  # Estatísticas descritivas do DataFrame
```

```python
print(df_funcionarios["Nome"])  # Acessa a coluna "Nome" do DataFrame
```

```python
print(df_funcionarios[["Nome", "Salario"]])  # Acessa as colunas "Nome" e "Salario" do DataFrame
```

```python
print(df_funcionarios[df_funcionarios["Salario"] > 4000])  # Filtra funcionários com salário maior que 4000
```

```python
print(df_funcionarios[df_funcionarios["Salario"] > 4000]["Nome"])  # Filtra e exibe apenas os nomes dos funcionários com salário maior que 4000
```

```python
resultado = df_funcionarios[(df_funcionarios["Salario"] > 4000) & 
                            (df_funcionarios["Setor"] == "TI")][["Nome", "Salario", "Setor"]]  # Filtra funcionários com salário maior que 4000 e setor "TI"
```

### Numpy

```python
import numpy as np
```

```python
notas = np.array([7.5, 8.0, 6.5, 9.0, 5.0])
```

```python
print("Notas:", notas)
```

```python
print("Média: ", np.mean(notas))  # Calcula a média das notas
print("Maior: ", np.max(notas))  # Calcula o valor máximo das notas
print("Mínimo: ", np.min(notas))  # Calcula o valor mínimo das notas
```

```python
print("+ 1 ponto", notas + 1)  # Adiciona 1 ponto a cada nota
```

```python
print("Aprovados:", notas >= 7)  # Filtra e exibe apenas as notas aprovadas (maiores ou iguais a 7)
```

```python
print("Notas Aprovadas: ", notas[notas >= 7])  # Exibe apenas as notas aprovadas (maiores ou iguais a 7)
```

```python
print("Soma: ", np.sum(notas))  # Calcula a soma de todas as notas
```
