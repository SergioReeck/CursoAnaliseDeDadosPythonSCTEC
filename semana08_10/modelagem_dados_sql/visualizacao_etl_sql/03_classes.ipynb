# Classes

Uma classe permite agrupar dados e funções que fazem parte do mesmo contexto.
Para Análise de Dados, isso pode ser útil para organizar códigos de tratamento, transformação e carga de dados.

Uma classe é como um modelo ou estrutura que define:

 - quais dados um objeto pode possuir;
 - quais ações esse objeto pode executar.

Uma analogia simples:

Uma classe pode ser vista como uma "forma" usada para criar objetos.

Para uma operação simples, uma função pode ser suficiente. Classes passam a ser interessantes quando queremos organizar dados e comportamentos relacionados.

### Exemplo:

````python
# Criando a classe
class Pessoa:

    def __init__(self, nome, idade):
        self.nome = nome
        self.idade = idade

# Criando o objeto
pessoa1 = Pessoa("João", 30)
pessoa2 = Pessoa("Maria", 25)

# Acessando os atributos
print(pessoa1.nome)
print(pessoa1.idade)
````

O __init__ é executado automaticamente quando criamos um objeto.

Exemplo: 

pessoa1 = Pessoa("João", 30)

Python executa o __init__ e atribui:
 - self.nome = "João"
 - self.idade = 30

 self representa o próprio objeto que está sendo utilizado.

### Métodos

Métodos são funções definidas dentro de uma classe.

````python
# Criar classe
class Pessoa:

    def __init__(self, nome):
        self.nome = nome

    def apresentar(self):
        print(f"Olá, meu nome é {self.nome}")
````

````python
# Criar objeto
pessoa = Pessoa("João")
````

````python
# Executar o método
pessoa.apresentar()
````

```text
# CLASSE
│
├── ATRIBUTOS
│   └── dados
│
└── MÉTODOS
    └── comportamentos / ações
```

## Exemplo: Televisao

Uma classe é um modelo que define quais informações um objeto possui e quais ações ele pode realizar.

Exemplo:

Televisão

 - Informações: ligada, canal, volume
 - Ações: ligar, desligar, mudar canal, aumentar volume

### TELEVISÃO

```text
             TELEVISÃO
                 │
       ┌─────────┴─────────┐
       ↓                   ↓
   ATRIBUTOS             MÉTODOS
       │                   │
   ligada              ligar()
   canal               desligar()
   volume              mudar_canal()
                       aumentar_volume()

````python
# Criar a classe
class Televisao:

    def __init__(self):
        self.ligada = False
        self.canal = 1
        self.volume = 10
````

 - ligada, canal e volume são atributos;
 - False, 1 e 10 são os valores iniciais;
 - __init__() define o estado inicial da televisão;
 - self representa a televisão que estamos criando.

````python
# Criar os métodos
class Televisao:

    def __init__(self):
        self.ligada = False
        self.canal = 1
        self.volume = 10

    def ligar(self):
        self.ligada = True

    def desligar(self):
        self.ligada = False

    def mudar_canal(self, canal):
        self.canal = canal
````

````python
# Criar o objeto
tv = Televisao()
````

````python
print(tv.ligada)
print(tv.canal)
````

````python
# Usar os métodos
tv.ligar()
tv.mudar_canal(5)
````

````python
print(tv.ligada)
print(tv.canal)
````

```text
Classe Televisao
       ↓
      tv
       ↓
┌─────────────────┐
│ ligada = True   │
│ canal = 5       │
│ volume = 10     │
└─────────────────┘
```
