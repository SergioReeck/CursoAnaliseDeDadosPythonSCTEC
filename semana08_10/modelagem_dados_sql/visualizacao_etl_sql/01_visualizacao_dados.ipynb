# Visualização de Dados

A visualização de dados permite transformar dados em gráficos,
facilitando a identificação de:

- tendências
- comparações
- distribuições
- relações entre variáveis
- padrões e possíveis problemas

Nesta aula vamos utilizar principalmente:

- Pandas: trabalha e organiza os dados
- Matplotlib: cria e personaliza gráficos
- Seaborn: facilita a criação de gráficos estatísticos e trabalha muito bem com DataFrames

## 1. Preparando o ambiente

Vamos importar as bibliotecas que serão utilizadas na aula.

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
````

## 2. Antes dos gráficos: uma pergunta importante

### Qual gráfico utilizar?

A escolha do gráfico depende da pergunta que queremos responder.

| Pergunta | Gráfico |
|---|---|
| Como algo evoluiu ao longo do tempo? | Linha |
| Qual categoria é maior? | Barras |
| Existe relação entre duas variáveis? | Dispersão |
| Como os valores estão distribuídos? | Histograma |
| Qual a participação de cada categoria? | Pizza |

Não existe um gráfico "melhor" para todos os casos.
O gráfico deve estar relacionado ao tipo de análise que queremos fazer.
"Que pergunta eu quero responder?"

## 3. Visualização de Dados com Matplotlib

### 1. Gráfico de linha — plt.plot()
- Ideal para tendências.
- Pergunta que responde: "Como as vendas evoluíram ao longo dos meses?"


````python
meses = ["Jan", "Fev", "Mar", "Abr", "Mai"]
vendas = [120, 150, 180, 160, 220]

plt.plot(meses, vendas)

plt.title("Vendas por mês")
plt.xlabel("Mês")
plt.ylabel("Vendas")

plt.show()
````

#### Personalização Básica

````python
plt.plot(
    meses,
    vendas,
    marker="o",
    linestyle="--"
)

plt.title("Vendas por mês")
plt.xlabel("Mês")
plt.ylabel("Quantidade de vendas")

plt.grid()
plt.show()
````

- marker="o" → mostra os pontos
- linestyle="--" → altera o estilo da linha
- grid() → adiciona uma grade

### 2. Gráfico de barras — plt.bar()
- Ideal para comparar categorias.
- Pergunta que responde: "Qual produto vendeu mais?"

````python
produtos = ["Notebook", "Celular", "Tablet", "Monitor"]
vendas = [35, 80, 45, 60]

plt.bar(produtos, vendas)

plt.title("Vendas por produto")
plt.xlabel("Produto")
plt.ylabel("Quantidade")

plt.show()
````

#### Gráfico de barras - plt.barh()
- Para gráficos de barra horizontal

````python
plt.barh(produtos, vendas)

plt.title("Vendas por produto")
plt.xlabel("Quantidade")
plt.ylabel("Produto")

plt.show()
````

### 3. Gráfico de dispersão — plt.scatter()
- Ideal para verificar relação entre duas variáveis numéricas.
- Pergunta que responde: "Existe relação entre idade e salário?"

````python
idade = [18, 22, 25, 30, 35, 40, 45]
salario = [1800, 2200, 2500, 3200, 4000, 5000, 6000]

plt.scatter(idade, salario)

plt.title("Idade x Salário")
plt.xlabel("Idade")
plt.ylabel("Salário")

plt.show()
````

#### Interpretação

Observe o gráfico.

- Os pontos parecem seguir algum padrão?
- Conforme a idade aumenta, o salário também aumenta?
- Podemos afirmar que uma variável causa a outra?

> Atenção: correlação não significa causalidade.
- Mesmo quando duas variáveis apresentam uma relação, isso não prova que uma variável está causando a outra.

### 4. Histograma — plt.hist()

O histograma mostra como os valores estão distribuídos.

Pergunta que responde:

> "Como os valores estão distribuídos?"

````python
idades = [18, 20, 21, 22, 23, 24, 25, 25, 26, 27,
          28, 29, 30, 31, 32, 35, 38, 40, 45, 50]

plt.hist(idades, bins=5)

plt.title("Distribuição das idades")
plt.xlabel("Idade")
plt.ylabel("Frequência")

plt.show()
````

### 5. Gráfico de pizza — plt.pie()

Pode ser utilizado para representar a participação de categorias
em relação a um total.

É mais adequado quando temos poucas categorias.

> Para muitas categorias, gráficos de barras geralmente facilitam
> mais a comparação.

````python
categorias = ["Eletrônicos", "Móveis", "Roupas", "Alimentos"]
vendas = [40, 25, 20, 15]

plt.pie(vendas, labels=categorias, autopct="%1.1f%%")

plt.title("Participação nas vendas")

plt.show()
````

# 4. Visualização de Dados com Pandas
> O Pandas utiliza o Matplotlib por baixo para gerar esses gráficos.

````python
# Criar DataFrame
df = pd.DataFrame({
    "produto": ["Notebook", "Celular", "Tablet", "Monitor"],
    "vendas": [35, 80, 45, 60]
})

df
````

````python
df.plot(
    x="produto",
    y="vendas",
    kind="bar"
)

plt.title("Vendas por produto")
plt.xlabel("Produto")
plt.ylabel("Quantidade")

plt.show()
````

#### kind
> O parâmetro kind na biblioteca Pandas do Python define o tipo de gráfico que você deseja criar ao usar a função .plot(). Ele serve para transformar dados de uma tabela (DataFrame) em uma imagem visual de forma rápida.

````python
df.plot(
    x="produto",
    y="vendas",
    kind="bar"
)
````

````python
# kind="line"
# kind="bar"
# kind="barh"
# kind="hist"
````

````python
df.plot(
    x="produto",
    y="vendas",
    kind="barh"
)

plt.show()
````

# 5. Visualização de Dados com Seaborn

O Seaborn é uma biblioteca de visualização construída sobre o Matplotlib.

Uma das suas principais vantagens é facilitar a criação de
gráficos estatísticos e trabalhar muito bem com DataFrames do Pandas.

````python
sns.barplot(
    data=df,
    x="produto",
    y="vendas"
)

plt.title("Vendas por produto")
plt.show()
````

````python
df_vendas = pd.DataFrame({
    "mes": ["Jan", "Jan", "Fev", "Fev", "Mar", "Mar",
            "Abr", "Abr", "Mai", "Mai"],
    "produto": [
        "Notebook", "Celular",
        "Notebook", "Celular",
        "Notebook", "Celular",
        "Notebook", "Celular",
        "Notebook", "Celular"
    ],
    "vendas": [30, 70, 40, 80, 45, 75, 50, 90, 60, 100]
})

df_vendas
````

**hue** cria uma divisão visual por categoria.

````python
sns.barplot(
    data=df_vendas,
    x="mes",
    y="vendas",
    hue="produto"
)

plt.title("Vendas por mês e produto")
plt.show()
````

### Seaborn — outros gráficos

O Seaborn possui diversos gráficos prontos para análise exploratória de dados.

Alguns exemplos:

| Gráfico | Função | Utilização |
|---|---|---|
| Barras | `sns.barplot()` | Comparar categorias |
| Dispersão | `sns.scatterplot()` | Analisar relação entre variáveis |
| Histograma | `sns.histplot()` | Visualizar distribuição dos dados |
| Linha | `sns.lineplot()` | Visualizar tendências ao longo do tempo |
| Boxplot | `sns.boxplot()` | Visualizar distribuição e possíveis outliers |
| Heatmap | `sns.heatmap()` | Visualizar relações/matrizes por intensidade de valores |

## Por que utilizar gráficos em Python?

Python possui diversas bibliotecas para visualização de dados, como:

- **Matplotlib**
- **Seaborn**

Essas bibliotecas permitem criar gráficos diretamente a partir dos dados e são muito úteis principalmente durante a **Análise Exploratória de Dados (EDA)**.

Por exemplo:

```python
plt.bar(produtos, vendas)
```

Com poucas linhas de código, podemos transformar dados em uma representação visual e analisar seu comportamento.

---

## Quem utiliza mais?

Tanto **Analistas de Dados** quanto **Cientistas de Dados** utilizam visualizações em Python, mas com objetivos diferentes.

### Analista de Dados

Utiliza gráficos principalmente para:

- Explorar os dados
- Identificar tendências e padrões
- Responder perguntas de negócio
- Apresentar resultados
- Apoiar decisões

### Cientista de Dados

Além disso, utiliza visualizações para:

- Explorar dados antes da modelagem
- Identificar relações entre variáveis
- Analisar distribuições
- Avaliar resultados de modelos
- Comunicar resultados de análises estatísticas e preditivas

> A visualização é importante para ambos, mas não substitui ferramentas especializadas de BI quando o objetivo é construir dashboards corporativos.

---

# Por que utilizar ferramentas de BI?

Python é excelente para **analisar, manipular e explorar dados**, mas existem ferramentas desenvolvidas especificamente para **visualização, dashboards e consumo de informações**.

Exemplos:

- **Power BI**
- **Qlik Sense**
- **Looker**
- **Tableau**
- **Amazon QuickSight**

Essas ferramentas oferecem recursos como:

- Dashboards interativos
- Filtros e segmentações
- Indicadores (KPIs)
- Interação entre gráficos
- Atualização automática dos dados
- Compartilhamento com usuários
- Controle de acesso
- Recursos voltados para usuários de negócio

## Em resumo

### Python

> **"Quero analisar e trabalhar com os dados."**

## Conclusão

Python possui excelentes recursos para criação de gráficos e é muito utilizado na **análise exploratória de dados**.
