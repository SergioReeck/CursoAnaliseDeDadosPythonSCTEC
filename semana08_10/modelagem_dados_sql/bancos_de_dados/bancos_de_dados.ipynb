# Introdução a Bancos de Dados com Python e SQLite

---

## O que é um Banco de Dados?

Um banco de dados é um sistema utilizado para armazenar, organizar e consultar informações.
Em vez de manter essas informações somente em arquivos como Excel ou CSV, podemos armazená-las em um banco de dados.

```text
CLIENTES

id    nome              cidade
-----------------------------------
1     João              Blumenau
2     Maria             Pomerode
3     Carlos            Gaspar

### Banco de Dados Relacional

Em um banco de dados relacional, os dados são organizados principalmente em tabelas.

```text
CLIENTES
-------------------------
id
nome
cidade


VENDAS
-------------------------
id
cliente_id
produto
valor

As tabelas podem se relacionar:

```text
CLIENTES
   │
   │ id
   ↓
VENDAS
   │
   │ produto
   ↓
PRODUTOS

**Exemplos de bancos relacionais**
 - PostgreSQL
 - MySQL
 - SQL Server
 - Oracle
 - SQLite

### Banco de Dados Não Relacional

Bancos não relacionais, também chamados de NoSQL, não dependem necessariamente do modelo tradicional de tabelas relacionais.

```json
{
    "id": 1,
    "nome": "João",
    "cidade": "Blumenau",
    "compras": [
        {
            "produto": "Notebook",
            "valor": 3500
        }
    ]
}

**Exemplos**
 - MongoDB
 - Redis
 - Cassandra
 - DynamoDB

---

## O que é SQL?

SQL — Structured Query Language é uma linguagem utilizada para trabalhar com bancos de dados relacionais.

### SQLite

Para esta aula vamos utilizar o SQLite.

O SQLite é um banco de dados relacional leve que funciona diretamente em um arquivo.

Não precisamos instalar um servidor de banco de dados para começar.

### Python + SQLite

O Python possui uma biblioteca própria para trabalhar com SQLite:
import sqlite3

---

## Utilizando o DB Browser for SQLite

Vamos utilizar inicialmente o **DB Browser for SQLite** para trabalhar com bancos de dados.

O DB Browser é uma ferramenta gráfica que permite criar, visualizar e manipular bancos de dados **SQLite** sem precisar executar os comandos diretamente pelo Python.

Neste primeiro momento, vamos utilizar a ferramenta para:

- Criar um banco de dados SQLite;
- Criar tabelas;
- Inserir dados;
- Visualizar os dados;
- Executar comandos SQL;
- Fazer consultas utilizando `SELECT`;
- Alterar e excluir registros.

Depois, vamos utilizar o **Python** para acessar esse mesmo banco de dados e executar comandos SQL programaticamente.

### Fluxo da aula

```text
DB Browser for SQLite
        ↓
   Banco SQLite
        ↓
 Criar tabelas
        ↓
 Inserir dados
        ↓
 Consultar dados
        ↓
       SQL
        ↓
      Python
        ↓
     Pandas
