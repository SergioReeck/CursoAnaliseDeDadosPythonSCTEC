# ETL - Extraindo e Manipulando Dados de API (IBGE)

### Consumindo Dados de uma API

```python
import requests

url = "https://servicodados.ibge.gov.br/api/v1/localidades/estados"

resposta = requests.get(url)
```

```python
print(resposta.status_code)
```

```python
estados = resposta.json()

print("-" * 30)

for estado in estados[:3]:
    print(estado)
```

### Gravando Dados em Arquivo JSON

```python
import json
```

```python
with open("estados.json", "w", encoding="utf-8") as arquivo:
    json.dump(estados, arquivo, indent=4, ensure_ascii=False)

print("Arquivo estados.json criado com sucesso!")
```

```python
# Verificando o conteúdo do arquivo estados.json

quantidade = 0

for estado in estados:
    quantidade += 1
print(f"Quantidade de estados: {quantidade}")
```

### Leitura do Arquivo JSON

```python
with open("estados.json", "r", encoding="utf-8") as arquivo:
    estados = json.load(arquivo)
```

### Convertendo JSON em DATAFRAME

```python
import pandas as pd
```

```python
df_estados = pd.DataFrame(estados)

print(df_estados.head())
```

### Selecinando Colunas

```python
# Selecionando apenas as colunas desejadas

df_estados = df_estados[["sigla", "nome", "regiao"]]

# Extraindo o nome da região de cada estado

df_estados["regiao"] = df_estados["regiao"].apply(lambda x: x["nome"])

df_estados.columns = ["UF", "Estado", "Regiao"]

print(df_estados)
```

### Gravando Dados em Excel

```python
df_estados.to_excel("estados.xlsx", index=False)

print("Arquivo estados.xlsx criado com sucesso!")
```
