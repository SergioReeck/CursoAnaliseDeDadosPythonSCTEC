```python
import re # expressões regulares REGEX
import pandas as pd
```

```python
dados = {"nome": ["Ana", "Carlos", "Maria", "José"],
        "telefone":["(47) 99999-1111", "47 98888-2222", "+55 (47) 97777-3333", "47966664444"]}
```

```python
df = pd.DataFrame(dados)
```

```python
df
```

```python
df_copy = df
```

```python
df["telefone"] = df["telefone"].apply(lambda x: re.sub(r"\D", "", x))
```

```python
df["telefone"] = df["telefone"].apply(lambda x: re.sub(r"[^0-9]", "", x))
```
