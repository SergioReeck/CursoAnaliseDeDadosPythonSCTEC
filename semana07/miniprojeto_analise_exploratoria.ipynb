# Validações e verificação dos valores

### Verificando valores nulos por coluna

```python
df.isnull().sum()
```

### Removendo espaços extras dos nomes das colunas

```python
df.columns.str.strip()

print(df.columns.tolist())
```

### Vericando valores incorretos

```python
valores_indefinidos = [
    "n/a",
    "n/d",
    "não informado",
    "não disponível",
    "indisponível",
]

contador = 0

for coluna in df.columns:
    for indice, valor in df[coluna].items():
        valor_normalizado = str(valor).strip().lower()

        if any(palavra in valor_normalizado for palavra in valores_indefinidos):
            contador += 1
            print(f"Encontrado na linha {indice}, coluna '{coluna}': {valor}")

print(f"Total de valores encontrados: {contador}")
```

### Verificando linhas duplicadas

```python
duplicadas = df[df.duplicated(keep="first")]
    
print("Quantidade de linhas duplicadas:", df.duplicated().sum())
```

### Verificando datas vazias

```python
print("Quantidade de datas vazias:", df["DATA"].isnull().sum())
```

### Verificando datas inválidas

```python
data_valida = r"^\d{2}/\d{2}/\d{4}$"

data_invalida = (
    df["DATA"].notna()
    & ~df["DATA"].astype(str).str.strip().str.match(data_valida, na=False)
)

print("Datas inválidas:")
print(df[data_invalida])
```

### Verificando valores vazios

```python
for coluna in df.columns:
    vazios = (
        df[coluna].isnull()
        | df[coluna].astype(str).str.strip().eq("")
    )

    quantidade_vazias = vazios.sum()

    if quantidade_vazias > 0:
        print(f"{coluna}: {quantidade_vazias} de linhas vazias")
    else:
        print(f"{coluna}: Nenhuma linha vazia encontrada")
```

# Tratamento e limpeza dos valores

### Preenchendo células com valores incorretos

```python
valores_indefinidos = [
    "n/a",
    "n/d",
    "não informado",
    "não disponível",
    "indisponível",
]

contador = 0

for coluna in df.columns:
    for indice, valor in df[coluna].items():
        valor_normalizado = str(valor).strip().lower()

        if any(palavra in valor_normalizado for palavra in valores_indefinidos):
            contador += 1
            df.loc[indice, coluna] = "SEM CATEGORIA"

print(f"Total de valores indefinidos encontrados: {contador}")
```

### Preenchendo células com valores nulos/vazios

```python
quantidade_vazias = df.isnull().sum().sum()

if quantidade_vazias > 0:
    df = df.fillna("SEM CATEGORIA")
    print("Quantidade de células preenchidas:", quantidade_vazias)
else:
    print("Nenhum valor vazio encontrado.")

# Obs.: Preferi preencher as colunas vazias do que excluí-las por inteiro por ser considerada uma prática pouco comum.
```

### Eliminando linhas duplicadas e salvando em um novo arquivo

```python
duplicadas = df[df.duplicated()]

print("Quantidade de linhas duplicadas eliminadas:", df.duplicated().sum())

df = df.drop_duplicates()

df.to_csv("base_varejo_limpa.csv", sep=";", index=False)
```

### Convertendo a coluna DATA de string para DATETIME

```python
df["DATA"] = pd.to_datetime(df["DATA"], dayfirst=True)

linhas_erro = df["DATA"].isnull().sum()

print(f"Quantidade de linhas com erro: {linhas_erro}")
print(f"Coluna 'DATA' convertida para o tipo: {df['DATA'].dtype}")
```
